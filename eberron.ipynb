````python
def extract_pdf_text(pdf_path: Path) -> str:
    """Extract the full text content from a PDF file."""
    reader = PdfReader(str(pdf_path))
    return "\n".join(page.extract_text() or "" for page in reader.pages)


# Build the list of document dictionaries, one per PDF file.
documents = []
for meta in DOCUMENT_METADATA:
    pdf_path = DOCUMENTS_DIR / meta["filename"]
    documents.append(
        {
            "title": meta["title"],
            "description": meta["description"],
            "content": extract_pdf_text(pdf_path),
            "filename": str(pdf_path),
        }
    )

print(f"Loaded {len(documents)} documents")
````

# Simple search with minsearch

Build a search index over the loaded documents with
[minsearch](https://github.com/alexeygrigorev/minsearch) and run a query.

- `text_fields`: fields indexed for full-text search (title, description, content)
- `keyword_fields`: fields usable for exact-match filtering (filename)
- `boost_dict`: optional per-field weights to tune ranking

````python
from minsearch import Index

# Index the loaded documents: full-text on title/description/content,
# filename kept as a keyword field for filtering.
index = Index(
    text_fields=["title", "description", "content"],
    keyword_fields=["filename"],
)
index.fit(documents)

print(f"Indexed {len(documents)} documents")
````

````python
# Run a simple search. boost_dict weights title/description matches higher
# than raw content matches; num_results limits the number of hits returned.
query = "tarius"
results = index.search(
    query,
    num_results=5,
    boost_dict={"title": 3.0, "description": 2.0, "content": 1.0},
)

for doc in results:
    print(f"Title:    {doc['title']}")
    print(f"Filename: {doc['filename']}")
    print("-" * 80)
````

# Chunking Documents for Better Retrieval

The document-level index retrieves whole documents, but some of them are huge
(the largest is over 1.3M characters). Sending 5 such documents to an LLM in a
RAG prompt would blow up the token count and dilute the answer.

Solution: split each large document into overlapping chunks with a **sliding window**
(via `gitsource.chunk_documents`), and index the chunks instead. Retrieval then
returns small, focused pieces of text.

Decisions for this notebook:

- `size=3000, step=1500` (50% overlap, so information split at boundaries is
  still fully contained in some chunk)
- The two small campaign-notes PDFs are **excluded from chunking** and added to
  the index as single whole-document chunks (`chunk_id=0`)
- One unified chunk index over all 10 documents

````python
# Analyze document sizes (largest first) to see why chunking is needed.
doc_sizes = [(doc["filename"], len(doc["content"])) for doc in documents]
doc_sizes.sort(key=lambda x: x[1], reverse=True)

for filename, size in doc_sizes[:5]:
    print(f"{filename}: {size} characters")
````

````python
from gitsource import chunk_documents

# The two campaign-notes PDFs are small enough to index whole.
EXCLUDE_FROM_CHUNKING = {
    "5208a1d4-62b5-495b-8905-382e6f8f946d.pdf",
    "Trascrizione Schede Personaggi D&D.pdf",
}

document_chunks = []
for doc in documents:
    filename = Path(doc["filename"]).name
    if filename in EXCLUDE_FROM_CHUNKING:
        # Whole-document chunk: keep all metadata, mark as the only chunk.
        document_chunks.append({**doc, "start": 0, "chunk_id": 0})
    else:
        # chunk_documents preserves metadata and adds 'start' + chunked 'content';
        # we add 'chunk_id' ourselves (it is not set by gitsource).
        for i, chunk in enumerate(chunk_documents([doc], size=3000, step=1500)):
            chunk["chunk_id"] = i
            document_chunks.append(chunk)

print(f"Created {len(document_chunks)} chunks from {len(documents)} documents")
````

````python
# Inspect one chunk: it carries the original document metadata
# plus 'start', 'chunk_id' and the chunked 'content'.
chunk = document_chunks[10]
{
    "start": chunk["start"],
    "content": chunk["content"][:150] + "...",
    "title": chunk["title"],
    "description": chunk["description"],
    "filename": chunk["filename"],
    "chunk_id": chunk["chunk_id"],
}
````

````python
# Chunks per document, and re-index the chunks with minsearch.
from collections import Counter

chunks_per_doc = Counter(Path(c["filename"]).name for c in document_chunks)
for filename, count in chunks_per_doc.most_common():
    print(f"{filename}: {count} chunks")

chunk_index = Index(
    text_fields=["title", "description", "content"],
    keyword_fields=["filename"],
)
chunk_index.fit(document_chunks)

print(f"\nChunk index fitted with {len(document_chunks)} chunks")
````

````python
# Demo query 1: rulebook content. We now get specific chunks
# instead of entire documents.
query = "warforged"
results = chunk_index.search(
    query,
    num_results=5,
    boost_dict={"title": 3.0, "description": 2.0, "content": 1.0},
)

for chunk in results:
    print(f"Title:    {chunk['title']}")
    print(f"Filename: {chunk['filename']}")
    print(f"Chunk:    #{chunk['chunk_id']} (start={chunk['start']})")
    print(f"Content:  {chunk['content'][:120]}...")
    print("-" * 80)
````

````python
# Demo query 2: campaign-notes-only content. This term exists only in the
# campaign diary PDF, proving the whole-document chunks are searchable.
query = "Profezia draconica"
results = chunk_index.search(
    query,
    num_results=5,
    boost_dict={"title": 3.0, "description": 2.0, "content": 1.0},
)

for chunk in results:
    print(f"Title:    {chunk['title']}")
    print(f"Filename: {chunk['filename']}")
    print(f"Chunk:    #{chunk['chunk_id']} (start={chunk['start']})")
    print("-" * 80)
````

# RAG: Search → Prompt → LLM

Complete documentation assistant:

1. **Search** - retrieve relevant chunks, then merge adjacent chunks from the
   same document into single passages
2. **Prompt** - build a prompt with the question and the (slimmed) search results
3. **LLM** - generate an answer grounded only in the retrieved context

Decisions for this notebook:

- Merging happens inside `search()`; the raw `chunk_index` remains available
- The prompt context contains only `{title, filename, content}` per passage
- The assistant answers in the language of the question (mixed EN/IT corpus)
- `num_results=3` merged passages to bound the prompt size

````python
import json
import os

from dotenv import load_dotenv
from openai import OpenAI

# Load OPENAI_API_KEY from the project's .env file.
load_dotenv()
assert os.environ.get("OPENAI_API_KEY"), "OPENAI_API_KEY not found - check your .env file"

openai_client = OpenAI()
print("OpenAI client ready")
````

````python
def merge_chunks(chunks):
    """Merge overlapping/adjacent chunks from the same document into one passage.

    Chunks know their absolute 'start' offset in the original document, so the
    overlap is written only once and the result is exact contiguous text.
    """
    chunks = sorted(chunks, key=lambda c: c["start"])
    merged = chunks[0]["content"]
    end = chunks[0]["start"] + len(merged)  # absolute end position in the document

    for chunk in chunks[1:]:
        chunk_end = chunk["start"] + len(chunk["content"])
        if chunk_end <= end:
            continue  # fully contained in what we already have
        overlap = max(end - chunk["start"], 0)
        if overlap == 0:
            merged += "\n...\n"  # gap between non-adjacent chunks
        merged += chunk["content"][overlap:]
        end = chunk_end

    return merged
````

````python
BOOST = {"title": 3.0, "description": 2.0, "content": 1.0}


def search(query, num_results=3):
    """Search chunks, then merge adjacent chunks per document into passages.

    Extra raw chunks are fetched because adjacent chunks from the same
    document collapse into a single merged passage.
    """
    hits = chunk_index.search(query, num_results=num_results * 3, boost_dict=BOOST)

    grouped = {}
    for hit in hits:
        grouped.setdefault(hit["filename"], []).append(hit)

    passages = []
    for filename, chunks in grouped.items():
        content = merge_chunks(chunks) if len(chunks) > 1 else chunks[0]["content"]
        passages.append(
            {
                "title": chunks[0]["title"],
                "filename": filename,
                "content": content,
            }
        )

    return passages[:num_results]
````

````python
def llm(user_prompt, instructions=None, model="gpt-4o-mini"):
    messages = []

    if instructions is not None:
        messages.append({"role": "system", "content": instructions})

    messages.append({"role": "user", "content": user_prompt})

    response = openai_client.responses.create(model=model, input=messages)

    return response.output_text
````

````python
instructions = """
You're a D&D campaign assistant for an Eberron campaign. Your task is to answer
the QUESTION using only the provided CONTEXT, which contains excerpts from D&D 5E
rulebooks and the campaign's own notes. If the context doesn't contain the
answer, say so. Cite which source (title/filename) each part of the answer
comes from. Always respond in the same language the QUESTION is written in,
regardless of the language of the CONTEXT.
"""


def build_prompt(query, search_results):
    search_result_json = json.dumps(search_results, indent=2)

    user_prompt = f"""
    <QUESTION>
    {query}
    </QUESTION>

    <CONTEXT>
    {search_result_json}
    </CONTEXT>
    """.strip()

    return user_prompt
````

````python
def rag(query):
    search_results = search(query)
    prompt = build_prompt(query, search_results)
    answer = llm(prompt, instructions)
    return answer
````

````python
# Demo 1: English rules question (rulebooks).
print(rag("Quale potrebbe essere il rapporto di Tarius e Zhull secondo gli appunti di campagna?"))
````
